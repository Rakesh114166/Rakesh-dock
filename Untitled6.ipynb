```python
# Cell 2: Run automated screening calculations and output results
import os
import pandas as pd
from colab_dock import ColabDockPipeline

# 1. Initialize your custom pipeline workspace folder name
pipeline = ColabDockPipeline(working_dir="phytochemical_screening")

# 2. Input the PDB ID of your target receptor protein
target_protein_id = "4EXS"
protein_file = pipeline.prepare_protein(pdb_id=target_protein_id)

# 3. Define your library (Add as many compounds or phytochemicals as you want!)
compounds_to_screen = {
    "Naringenin": "O=C1CC(C2=CC=C(O)C=C2)OC3=CC(O)=CC(O)=C13",
    "Chrysin": "O=C1C=C(C2=CC=CC=C2)OC3=CC(O)=CC(O)=C13"
}

# 4. Set the exact active/binding site coordinates of your target pocket
pocket_center = [12.5, -3.2, 24.1]
pocket_size = [20.0, 20.0, 20.0]

results_list = []

print("\n▶️ Launching automated virtual screening calculation loop...")
for name, smiles in compounds_to_screen.items():
    try:
        ligand_file = pipeline.prepare_ligand(smiles, name)
        affinity = pipeline.run_docking(
            receptor_pdbqt=protein_file,
            ligand_pdbqt=ligand_file,
            center=pocket_center,
            size=pocket_size,
            output_name=name,
            exhaustiveness=16
        )
        results_list.append({"Compound ID": name, "Binding Affinity (kcal/mol)": affinity})
        print(f"   ✅ Processing complete for {name}: {affinity} kcal/mol")
    except Exception as e:
        print(f"   ❌ Skipping compound {name} due to calculation error: {e}")

# 5. Display the final evaluation metrics matrix table
if results_list:
    df = pd.DataFrame(results_list).sort_values(by="Binding Affinity (kcal/mol)").reset_index(drop=True)
    print("\n" + "="*50)
    print("📊 FINAL RANKED VIRTUAL SCREENING RESULTS:")
    print("="*50)
    display(df)
```
