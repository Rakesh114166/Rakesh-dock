```python
# Cell 2: Run automated screening calculations with automated pocket centering
import os
import pandas as pd
from vina import Vina

# Helper function to automatically calculate the exact geometric center of the protein
def calculate_protein_center(pdbqt_path):
    xs, ys, zs = [], [], []
    with open(pdbqt_path, 'r') as f:
        for line in f:
            if line.startswith("ATOM") or line.startswith("HETATM"):
                try:
                    xs.append(float(line[30:38]))
                    ys.append(float(line[38:46]))
                    zs.append(float(line[46:54]))
                except ValueError:
                    continue
    if not xs:
        return [0.0, 0.0, 0.0]
    return [round(sum(xs)/len(xs), 2), round(sum(ys)/len(ys), 2), round(sum(zs)/len(zs), 2)]

class ColabDockPipeline:
    def __init__(self, working_dir="docking_project"):
        self.working_dir = working_dir
        self.prot_dir = os.path.join(working_dir, "protein_data")
        self.lig_dir = os.path.join(working_dir, "ligand_data")
        os.makedirs(self.prot_dir, exist_ok=True)
        os.makedirs(self.lig_dir, exist_ok=True)

    def prepare_protein(self, pdb_id):
        """Downloads a target PDB structure and converts it cleanly to PDBQT via command-line OpenBabel."""
        pdb_path = os.path.join(self.prot_dir, f"{pdb_id}.pdb")
        pdbqt_path = os.path.join(self.prot_dir, f"{pdb_id}_receptor.pdbqt")
        os.system(f"wget -q -O {pdb_path} https://files.rcsb.org/download/{pdb_id}.pdb")

        cmd = f"obabel {pdb_path} -O {pdbqt_path} -h -xr -d --quiet"
        os.system(cmd)
        return pdbqt_path

    def prepare_ligand(self, smiles, ligand_name):
        """Generates 3D coordinates directly from SMILES lines and parameters bonds via OpenBabel."""
        pdbqt_out = os.path.join(self.lig_dir, f"{ligand_name}.pdbqt")
        cmd = f'obabel -:"{smiles}" -O {pdbqt_out} --gen3d -h -p 7.4 --quiet'
        os.system(cmd)
        return pdbqt_out

    def run_docking(self, receptor_pdbqt, ligand_pdbqt, center, size, output_name, exhaustiveness=16):
        """Executes native screening calculations using the AutoDock Vina engine."""
        output_pose = os.path.join(self.lig_dir, f"{output_name}_poses.pdbqt")
        v = Vina(sf_name='vina')
        v.set_receptor(receptor_pdbqt)
        v.set_ligand_from_file(ligand_pdbqt)

        v.compute_vina_maps(center=center, box_size=size)
        v.dock(exhaustiveness=exhaustiveness, n_poses=9)
        v.write_poses(output_pose, n_poses=9, overwrite=True)
        return v.energies(n_poses=1)[0][0]

# Initialize pipeline workspace mapping parameters
pipeline = ColabDockPipeline(working_dir="virtual_screening_project")
protein_file = pipeline.prepare_protein(pdb_id="4EXS")

# AUTOMATION STEP: Dynamically find the real center of the protein structure
pocket_center = calculate_protein_center(protein_file)
print(f"🎯 Automated box centering coordinates calculated: {pocket_center}")

# Define searching dimensions (large enough to wrap around the target pocket site)
pocket_size = [25.0, 25.0, 25.0]

compounds_to_screen = {
    "Naringenin": "O=C1CC(C2=CC=C(O)C=C2)OC3=CC(O)=CC(O)=C13",
    "Chrysin": "O=C1C=C(C2=CC=CC=C2)OC3=CC(O)=CC(O)=C13"
}

results_list = []
print("\n▶️ Launching automated virtual screening calculation loop...")
for name, smiles in compounds_to_screen.items():
    try:
        ligand_file = pipeline.prepare_ligand(smiles, name)
        affinity = pipeline.run_docking(
            receptor_pdbqt=protein_file,
            ligand_pdbqt=ligand_file,
            center=pocket_center,
            size=pocket_size,
            output_name=name,
            exhaustiveness=8 # Quick testing speed
        )
        results_list.append({"Compound ID": name, "Binding Affinity (kcal/mol)": affinity})
        print(f"   ✅ Processing complete for {name}: {affinity} kcal/mol")
    except Exception as e:
        print(f"   ❌ Skipping compound {name} due to calculation error: {e}")

# Display complete sorted evaluation summary metrics matrix table
if results_list:
    df = pd.DataFrame(results_list).sort_values(by="Binding Affinity (kcal/mol)").reset_index(drop=True)
    print("\n" + "="*50)
    print("📊 FINAL RANKED VIRTUAL SCREENING RESULTS:")
    print("="*50)
    display(df)
```
